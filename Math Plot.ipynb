```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib as mpl
```

```python
fig = plt.figure() # an empty figure, no axes
fig, ax = plt.subplots(2,3) # a grid of plots, 2 rows, 3 columns
```

```python
x = np.linspace(-2,2,100) 
plt.plot(x, x**2, label='parabola') # plot y = x^2
plt.plot(x, x**3, label='cubic') # plot y = x^3
plt.xlabel('something funny')
plt.ylabel('something funnier')
plt.title('A funny plot')
plt.legend()
```
