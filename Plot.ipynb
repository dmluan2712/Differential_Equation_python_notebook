```python
import numpy as np
import itertools
from matplotlib.backends.backend_pdf import PdfPages
import matplotlib.pyplot as plt
Y,X = np.mgrid[-2:2:.01, -2:2:.01]
y,x = Y[:,0], X[0,:]
U, V = X+Y, -5*X-Y
stream_points = np.array(list(itertools.product(np.arange(-1.9,1.9,0.75), np.arange(-1.9,1.9,0.75))))
plt.streamplot(x,y, U,V, start_points=stream_points,  broken_streamlines=False, density=1);
#plt.plot(x,x,color='red')
#plt.plot(x,-x,color='red')
plt.xlim(-2,2)
plt.ylim(-2,2)
plt.gca().set_aspect('equal')
plt.savefig('PP-1a.pdf')
```
